```python
from scipy import stats

df_anova = df.dropna(subset=["Yield_Tonnes_Ha"])

seasons = df_anova["Season"].unique()

groups = []

for season in seasons:
    group = df_anova[df_anova["Season"] == season]["Yield_Tonnes_Ha"]
    groups.append(group)

result = stats.f_oneway(*groups)

print("F-statistic:", result.statistic)
print("P-value:", result.pvalue)

if result.pvalue < 0.05:
    print("There is a significant difference in yield between seasons.")
else:
    print("There is no significant difference in yield between seasons.")
```

```python
yield_season = df.groupby("Season")["Yield_Tonnes_Ha"].mean()

best_season = yield_season.idxmax()
best_season_yield = yield_season.max()

print("Best season:", best_season)
print("Average yield:", best_season_yield)
```

```python
crop_yield = df.groupby("Crop")["Yield_Tonnes_Ha"].mean()

best_crop = crop_yield.idxmax()
best_crop_yield = crop_yield.max()

print("Best crop:", best_crop)
print("Average yield:", best_crop_yield)
```

```python
profit_season = df.groupby("Season")["Profit_INR"].mean()

best_profit_season = profit_season.idxmax()

best_profit = profit_season.max()

print("Most profitable season:", best_profit_season)
print("Average profit:", best_profit)
```

```python
risk_season = df.groupby("Season")["Disease_Pest_Risk_pct"].mean()

print(risk_season)

risk_season.plot(kind="bar")

plt.title("Average Disease and Pest Risk Across Seasons")
plt.xlabel("Season")
plt.ylabel("Risk (%)")

plt.show()
```

## Key Findings

* The dataset contains 4,000 agricultural records.
* **Kharif** had the highest average yield at about **5.64 Tonnes/Ha**.
* **Kharif** also had the highest average profit of about **₹1.79 lakh**.
* **Sugarcane** had the highest average crop yield at about **46.94 Tonnes/Ha**.
* Rainfall, temperature, water, and fertilizer usage showed differences across seasons.


## Interpretation

Overall, **Kharif performed the best** in terms of both yield and profit. The analysis also shows that environmental and resource-related factors can vary between seasons and are related to agricultural performance.


## Recommendations

* Plan crops according to seasonal conditions.
* Use water and fertilizers efficiently.
* Monitor rainfall and temperature when planning irrigation.
* Keep track of disease and pest risks.
* Consider both yield and profit when choosing crops.


## Conclusion

This project helped identify seasonal patterns in agricultural performance. **Kharif showed the best overall yield and profit**, while Sugarcane had the highest average crop yield. These findings can help with better crop planning and resource management.
